```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
iris_data = pd.read_csv("IRIS.csv")
```

```python
print(iris_data.head())
```

```python
print(iris_data.shape)
```

```python
print(iris_data.info())
```

```python
print(iris_data.isnull().sum())
```

```python
sns.pairplot(iris_data, hue='species')

plt.show()
```

```python
sns.countplot(x='species', data=iris_data)
plt.title("Species Count")
plt.show()
```

```python
numeric_iris_data = iris_data.copy()
numeric_iris_data['species'] = numeric_iris_data['species'].astype('category').cat.codes
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(numeric_iris_data.corr(),annot=True,cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=iris_data.drop('species', axis=1))
plt.title("Feature Distribution")
plt.show()
```

```python
encoder = LabelEncoder()
iris_data['species'] = encoder.fit_transform(iris_data['species'])
```

```python
print(iris_data.head())
```

```python
X = iris_data.drop('species', axis=1)
y = iris_data['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=7)
```

```python
model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy*100)
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(confusion_matrix(y_test, y_pred))
```

```python
sample = [[5.1, 3.5, 1.4, 0.2]]
prediction = model.predict(sample)
print("Predicted Species:", encoder.inverse_transform(prediction))
```

```python
new_data = pd.DataFrame([[5.1,3.5,1.4,0.2]],
columns=['sepal_length','sepal_width','petal_length','petal_width'])
prediction = model.predict(new_data)
print(prediction)
```

```python
from sklearn.svm import SVC
model_svc = SVC()
model_svc.fit(X_train, y_train)
```

```python
prediction1 = model_svc.predict(X_test)
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test, prediction1)*100)
```

```python
from sklearn.linear_model import LogisticRegression
model_lr = LogisticRegression()
model_lr.fit(X_train, y_train)
```

```python
prediction2 = model_lr.predict(X_test)
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test, prediction2)*100)
```
